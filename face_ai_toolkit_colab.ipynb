## 🔄 Face Swap

Swap faces between two images:

```python
# Face Swap Example
source_image = 'uploads/source.jpg'  # Change to your source image
target_image = 'uploads/target.jpg'  # Change to your target image
output_path = 'outputs/face_swapped.jpg'

try:
    # Perform face swap
    result = face_ai.face_swap(source_image, target_image, output_path)
    
    # Display results
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    # Source
    source = cv2.imread(source_image)
    source = cv2.cvtColor(source, cv2.COLOR_BGR2RGB)
    axes[0].imshow(source)
    axes[0].set_title('Source')
    axes[0].axis('off')
    
    # Target
    target = cv2.imread(target_image)
    target = cv2.cvtColor(target, cv2.COLOR_BGR2RGB)
    axes[1].imshow(target)
    axes[1].set_title('Target')
    axes[1].axis('off')
    
    # Result
    result_img = cv2.imread(output_path)
    result_img = cv2.cvtColor(result_img, cv2.COLOR_BGR2RGB)
    axes[2].imshow(result_img)
    axes[2].set_title('Face Swapped')
    axes[2].axis('off')
    
    plt.tight_layout()
    plt.show()
    
    print("✅ Face swap completed!")
    
except Exception as e:
    print(f"❌ Face swap failed: {e}")
    print("Make sure you have uploaded source.jpg and target.jpg")
```

## ✨ Face Enhancement

Enhance face quality using GFPGAN:

```python
# Face Enhancement Example
input_image = 'uploads/low_quality.jpg'  # Change to your input image
output_path = 'outputs/enhanced_face.jpg'

try:
    # Enhance face
    result = face_ai.enhance_face(input_image, output_path)
    
    # Display results
    fig, axes = plt.subplots(1, 2, figsize=(12, 6))
    
    # Original
    original = cv2.imread(input_image)
    original = cv2.cvtColor(original, cv2.COLOR_BGR2RGB)
    axes[0].imshow(original)
    axes[0].set_title('Original')
    axes[0].axis('off')
    
    # Enhanced
    enhanced = cv2.imread(output_path)
    enhanced = cv2.cvtColor(enhanced, cv2.COLOR_BGR2RGB)
    axes[1].imshow(enhanced)
    axes[1].set_title('Enhanced')
    axes[1].axis('off')
    
    plt.tight_layout()
    plt.show()
    
    print("✅ Face enhancement completed!")
    
except Exception as e:
    print(f"❌ Face enhancement failed: {e}")
    print("Make sure you have uploaded an image file")
```

## 🔍 Frame Enhancement

Upscale images using Real-ESRGAN:

```python
# Frame Enhancement Example
input_image = 'uploads/low_res.jpg'  # Change to your input image
output_path = 'outputs/upscaled_4x.jpg'

try:
    # Enhance frame (4x upscaling)
    result = face_ai.enhance_frame(input_image, output_path, scale=4)
    
    # Display results
    fig, axes = plt.subplots(1, 2, figsize=(15, 8))
    
    # Original
    original = cv2.imread(input_image)
    original = cv2.cvtColor(original, cv2.COLOR_BGR2RGB)
    axes[0].imshow(original)
    axes[0].set_title(f'Original ({original.shape[1]}x{original.shape[0]})')
    axes[0].axis('off')
    
    # Upscaled
    upscaled = cv2.imread(output_path)
    upscaled = cv2.cvtColor(upscaled, cv2.COLOR_BGR2RGB)
    axes[1].imshow(upscaled)
    axes[1].set_title(f'Upscaled 4x ({upscaled.shape[1]}x{upscaled.shape[0]})')
    axes[1].axis('off')
    
    plt.tight_layout()
    plt.show()
    
    print("✅ Frame enhancement completed!")
    
except Exception as e:
    print(f"❌ Frame enhancement failed: {e}")
    print("Make sure you have uploaded an image file")
```

## 👴 Age Transformation

Transform face to different ages:

```python
# Age Transformation Example
input_image = 'uploads/person.jpg'  # Change to your input image

# Transform to different ages
ages = [25, 45, 65]
results = []

try:
    for age in ages:
        output_path = f'outputs/aged_{age}.jpg'
        face_ai.age_transform(input_image, output_path, target_age=age)
        results.append(output_path)
    
    # Display results
    fig, axes = plt.subplots(1, len(ages) + 1, figsize=(20, 5))
    
    # Original
    original = cv2.imread(input_image)
    original = cv2.cvtColor(original, cv2.COLOR_BGR2RGB)
    axes[0].imshow(original)
    axes[0].set_title('Original')
    axes[0].axis('off')
    
    # Age transformations
    for i, (age, result_path) in enumerate(zip(ages, results)):
        aged = cv2.imread(result_path)
        aged = cv2.cvtColor(aged, cv2.COLOR_BGR2RGB)
        axes[i + 1].imshow(aged)
        axes[i + 1].set_title(f'Age {age}')
        axes[i + 1].axis('off')
    
    plt.tight_layout()
    plt.show()
    
    print("✅ Age transformation completed!")
    
except Exception as e:
    print(f"❌ Age transformation failed: {e}")
    print("Make sure you have uploaded a person image")
```

## 👄 Lip Sync

Synchronize lip movements with audio:

```python
# Lip Sync Example
video_file = 'uploads/person_video.mp4'  # Change to your video file
audio_file = 'uploads/speech.wav'        # Change to your audio file
output_path = 'outputs/lip_synced.mp4'

try:
    # Perform lip sync
    result = face_ai.lip_sync(video_file, audio_file, output_path)
    
    print("✅ Lip sync completed!")
    print(f"📹 Output saved to: {output_path}")
    
    # Display video info
    cap = cv2.VideoCapture(output_path)
    fps = cap.get(cv2.CAP_PROP_FPS)
    frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    duration = frame_count / fps
    
    print(f"📊 Video Info:")
    print(f"   Duration: {duration:.2f} seconds")
    print(f"   FPS: {fps}")
    print(f"   Frames: {frame_count}")
    
    cap.release()
    
except Exception as e:
    print(f"❌ Lip sync failed: {e}")
    print("Make sure you have uploaded both video and audio files")
```

## 🔄 Batch Processing

Process multiple files at once:

```python
# Batch Enhancement Example
from face_ai_toolkit import FaceEnhancer

enhancer = FaceEnhancer(device='cuda' if torch.cuda.is_available() else 'cpu')

try:
    # Batch enhance all images in uploads folder
    enhancer.batch_enhance('uploads/', 'outputs/batch_enhanced/')
    
    print("✅ Batch processing completed!")
    
    # List processed files
    if os.path.exists('outputs/batch_enhanced/'):
        processed_files = os.listdir('outputs/batch_enhanced/')
        print(f"📋 Processed {len(processed_files)} files:")
        for file in processed_files:
            print(f"   - {file}")
    
except Exception as e:
    print(f"❌ Batch processing failed: {e}")
```

## 💡 Tips & Troubleshooting

**Performance Tips:**
- Use T4 GPU for 10x faster processing
- Process images in batches for efficiency
- Clear GPU cache between large operations

**Common Issues:**
- **Out of Memory**: Reduce image size or use CPU
- **Model Not Found**: Re-run model download cell
- **No Faces Detected**: Ensure clear face visibility

**File Requirements:**
- Images: JPG, PNG (max 2048x2048 recommended)
